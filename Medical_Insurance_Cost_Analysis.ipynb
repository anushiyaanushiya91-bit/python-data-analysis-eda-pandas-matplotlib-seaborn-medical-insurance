**Dataset Size:** 1,338 records and 7 features.

**Year / Timeline:** The dataset does not specify a particular data collection year or timeline.

## Dataset Column / Features Description

* **Age** – It represents the age of the primary insurance beneficiary.

* **Sex** – It represents the gender of the insurance beneficiary, such as male or female.

* **BMI** – It represents the Body Mass Index of the beneficiary and indicates the individual's body weight relative to height.

* **Children** – It represents the number of children or dependents covered by the health insurance.

* **Smoker** – It indicates whether the insurance beneficiary is a smoker or non-smoker.

* **Region** – It represents the residential region of the insurance beneficiary, such as northeast, northwest, southeast, or southwest.

* **Charges** – It represents the individual medical costs billed by the health insurance provider. This is the main target variable used to analyse medical insurance costs.

**MULTIVARIATE ANALYSIS**:
**AGE, BMI, AND SMOKER STATUS VS INSURANCE CHARGES**

```python
sns.pairplot(
    df,
    vars=['age', 'bmi', 'charges'],
    hue='smoker'
)

plt.suptitle(
    'Multivariate Analysis: Age, BMI and Smoker Status vs Insurance Charges',
    y=1.02
)

plt.show()
```

**MULTI VARIATE ANALYSIS -GROUPED ANALYSIS**

```python
grouped_analysis = df.groupby(
    ['smoker', 'sex']
)['charges'].mean().reset_index()

print(grouped_analysis)
```

**INTERPRETATION**

Average insurance charges differ between smokers and non-smokers.
The grouped analysis also compares average charges across genders.
Smoking status may have a stronger relationship with charges than gender.

**BUSINESS INSIGHT**

Insurance companies can compare average insurance costs by smoking status and gender.

**PIVOT TABLES**

```python
pivot_analysis = df.pivot_table(
    values='charges',
    index='smoker',
    columns='sex',
    aggfunc='mean'
)

print(pivot_analysis)
```

**INTERPRETATION**

The pivot table compares average insurance charges across smoking status and gender.
Differences between groups help identify patterns in insurance costs.

**BUSINESS INSGIHTS**

Insurance companies can use group-level comparisons to understand customer risk patterns.
Smoking status and gender can be evaluated together when analysing insurance costs.
